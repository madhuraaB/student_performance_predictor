```python
import numpy as np 
import pandas as pd
import random
```

```python
df = pd.read_csv(r"D:\student_performance_predictor\dataset\dataset.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())
```

```python
df = df.drop_duplicates()
```

```python
print("After duplicates:", df.shape)
```

```python
strong_student = (
    (df["Attendance"] >= 90) &
    (df["Previous_Scores"] >= 90) &
    (df["Hours_Studied"] >= 20)
)

weak_result = (
    df["Exam_Score"] < 70
)
before = len(df)
df = df[~(strong_student & weak_result)]
print("Contradictions removed:", before - len(df))
```

```python
weak_profile = (
    (df["Hours_Studied"] < 10) &
    (df["Previous_Scores"] < 60)
)

top_score = (
    df["Exam_Score"] >= 90
)
```

```python
before = len(df)

df = df[~(weak_profile & top_score)]

print("Weak toppers removed:", before - len(df))
```

```python
# After all cleaning steps

toppers = df[df["Exam_Score"] >= 85]

print("Original shape:", df.shape)
print("Topper rows:", toppers.shape)

# Duplicate toppers once
df = pd.concat(
    [df, toppers],
    ignore_index=True
)

print("New shape:", df.shape)
```
