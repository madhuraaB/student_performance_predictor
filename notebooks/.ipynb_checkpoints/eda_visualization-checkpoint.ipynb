```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("D:/student_performance_predictor/dataset/cleaned_dataset.csv")
```

```python
df.head()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Exam_Score"], bins=20)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Attendance"], df["Exam_Score"])

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance")
plt.ylabel("Exam Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Hours_Studied"],df["Exam_Score"])
plt.title("Hours Studied vs Exam score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Previous_Scores"],df["Exam_Score"])
plt.title("Previous Score vs Exam Score")
plt.xlabel("Previous Score")
plt.ylabel("Exam Score")
plt.show()
```

```python
encoded_df = pd.get_dummies(df, drop_first=True)

# Correlation with Exam Score
corr_with_score = encoded_df.corr()["Exam_Score"]

# Remove Exam_Score itself
corr_with_score = corr_with_score.drop("Exam_Score")

# Sort values
corr_with_score = corr_with_score.sort_values(ascending=False)

# Convert to dataframe
corr_df = corr_with_score.to_frame()

# Plot
plt.figure(figsize=(8,10))

sns.heatmap(
    corr_df,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Feature Correlation with Exam Score")

plt.show()
```
